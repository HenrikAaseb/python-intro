## 7.1 Hva er et bibliotek?

Et **bibliotek** (også kalt en modul eller pakke) er en samling av ferdigskrevet kode som andre har laget, som vi kan bruke i våre egne programmer. I stedet for å finne opp hjulet på nytt hver gang, kan vi importere funksjoner som allerede finnes.

Python har mange innebygde biblioteker, og enda flere som kan installeres. Noen vi kommer til å bruke:
- `random` – for tilfeldige tall
- `numpy` – for matematiske beregninger og arrayer
- `matplotlib` (via `pylab`) – for å tegne grafer (neste notebook!)

### Måter å importere på

```python
import random              # importerer hele biblioteket - bruk random.randint(...)
import random as r         # gir biblioteket et kortere kallenavn - bruk r.randint(...)
from random import randint # importerer bare én funksjon - bruk randint(...) direkte
from random import *       # importerer ALT fra biblioteket direkte
```

````python
import random

tall = random.randint(1, 6)   # Tilfeldig heltall mellom 1 og 6 (som en terning)
print(tall)
````

## 7.2 random-biblioteket

`random`-biblioteket gir oss flere nyttige funksjoner for å lage tilfeldige verdier.

| Funksjon | Beskrivelse | Eksempel |
|---|---|---|
| `randint(a, b)` | Tilfeldig **heltall** fra og med `a` til og med `b` | `randint(1, 6)` |
| `uniform(a, b)` | Tilfeldig **flyttall** mellom `a` og `b` | `uniform(1, 6)` |
| `choice(liste)` | Trekker et tilfeldig element fra en liste | `choice(["a", "b", "c"])` |

````python
from random import randint, uniform, choice

print(randint(1, 100))                       # Tilfeldig heltall mellom 1 og 100
print(uniform(0, 1))                          # Tilfeldig flyttall mellom 0 og 1
print(choice(["stein", "saks", "papir"]))     # Tilfeldig valg fra en liste
````

### 💡 Eksempel: Simulere terningkast

````python
from random import randint

terning_1 = randint(1, 6)
terning_2 = randint(1, 6)

print("Terning 1:", terning_1)
print("Terning 2:", terning_2)

if terning_1 > terning_2:
    print("Terning 1 vant!")
elif terning_2 > terning_1:
    print("Terning 2 vant!")
else:
    print("Uavgjort!")
````

## 7.3 Nyttige innebygde funksjoner

Python har flere innebygde funksjoner som er nyttige å kjenne til når du jobber med tall og lister. Disse trenger ikke noen import.

````python
tall_liste = [4, 12, 7, 25, 3]

print(len(tall_liste))     # Antall elementer i listen
print(sum(tall_liste))     # Summen av alle elementene
print(max(tall_liste))     # Det høyeste tallet
print(min(tall_liste))     # Det laveste tallet
print(round(3.777, 2))     # Avrunder til 2 desimaler
print(abs(-15))            # Absoluttverdien (alltid positiv)
````

### 🔧 Oppgave 7.3.1

- Antall karakterer i listen
- Gjennomsnittet (summen delt på antallet)

## 7.4 numpy-biblioteket

`numpy` (Numerical Python) er et bibliotek for matematiske beregninger, og brukes ofte sammen med arrays (se forrige notebook). Det gir oss også nyttige matematiske funksjoner og konstanter.

````python
import numpy as np

print(np.pi)              # Verdien til pi (π)
print(np.sqrt(64))         # Kvadratrot
print(np.round(3.14159, 2))  # Avrunding

liste_tall = np.array([1, 2, 3, 4, 5])
print(np.sum(liste_tall))     # Sum av elementene i arrayet
print(liste_tall * 2)         # Ganger alle elementene med 2
````

Lag et program som simulerer at du kaster en mynt (krone eller mynt) 20 ganger. Bruk `randint(0, 1)` der `0` betyr kron og `1` betyr mynt. Tell opp og skriv ut hvor mange ganger du fikk hver side.

**Oppgave D**  
Lag et program som trekker 5 tilfeldige heltall mellom 1 og 100, legger dem i en liste, og bruker `sorted()` eller `.sort()` til å skrive dem ut i stigende rekkefølge.

````python
# Oppgave C
from random import randint

kron = 0
mynt = 0
for _ in range(20):
    side = randint(0, 1)
    if side == 0:
        kron += 1
    else:
        mynt += 1

print("Kron:", kron)
print("Mynt:", mynt)
````

### 🔴 Vanskelig nivå (utfordrende)

**Oppgave E**  
Lag et program som simulerer 10 000 terningkast med en `for`-løkke, og skriver ut hvor mange ganger hver verdi (1-6) kom opp. Er fordelingen omtrent lik for alle verdiene? Hvorfor/hvorfor ikke?

**Oppgave F**  
Lag en enkel "Stein, saks, papir"-motstander: la datamaskinen velge tilfeldig med `choice()`, spør brukeren om deres valg med `input()`, og skriv ut hvem som vant.

````python
# Oppgave E
from random import randint

antall = [0, 0, 0, 0, 0, 0]
for _ in range(10_000):
    terningkast = randint(1, 6)
    antall[terningkast - 1] += 1

for verdi in range(1, 7):
    print(f"{verdi}: {antall[verdi - 1]} ganger")

print("Fordelingen blir omtrent jevn, men ikke helt lik, fordi kastene er tilfeldige.")
````

````python
# Oppgave F
from random import choice

valg = ["stein", "saks", "papir"]
datamaskin = choice(valg)
bruker = input("Velg stein, saks eller papir: ").strip().lower()

while bruker not in valg:
    bruker = input("Ugyldig valg. Skriv stein, saks eller papir: ").strip().lower()

print("Datamaskinen valgte:", datamaskin)
if bruker == datamaskin:
    print("Uavgjort!")
elif (bruker == "stein" and datamaskin == "saks") or \
     (bruker == "saks" and datamaskin == "papir") or \
     (bruker == "papir" and datamaskin == "stein"):
    print("Du vant!")
else:
    print("Datamaskinen vant!")
````
